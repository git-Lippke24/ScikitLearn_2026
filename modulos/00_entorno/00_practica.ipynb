## TODO 1 · Verificar el entorno

Lo primero es confirmar que el notebook está usando el Python correcto y que las
librerías instaladas cumplen los mínimos del plan.

````python
# === TODO 1 — Verificar el interprete y las versiones ===
#
# Que hay que hacer:
#   a) Guardar en `ejecutable` la ruta del Python que esta corriendo este notebook.
#   b) Guardar en `version_sklearn` la version instalada de scikit-learn, como string.
#
# Por que:
#   Si el kernel apunta a un Python distinto del .venv, tendras las librerias
#   instaladas y aun asi un ModuleNotFoundError. Es el error mas comun al empezar
#   y el mas desconcertante, porque `pip list` en la terminal se ve perfecto.
#
# Que debe quedar:
#   ejecutable      -> str, la ruta absoluta al interprete
#   version_sklearn -> str, por ejemplo "1.5.2"
#
# Pistas: el modulo `sys` tiene un atributo con la ruta del ejecutable.
#         Casi toda libreria expone su version en `__version__`.

import sys
import sklearn

# --- tu codigo aqui ---
ejecutable = sys.executable   # Para saber que python se está usando.
version_sklearn = sklearn.__version__  # Para saber la version de scikit-learn. 


# --- verificacion (no modificar) ---
assert isinstance(ejecutable, str), "ejecutable debe ser un string"
assert isinstance(version_sklearn, str), "version_sklearn debe ser un string"

mayor, menor = (int(p) for p in version_sklearn.split(".")[:2])
assert (mayor, menor) >= (1, 5), (
    f"Este curso necesita scikit-learn >= 1.5 y tienes {version_sklearn}. "
    "Ejecuta: pip install -U scikit-learn"
)

print("interprete       :", ejecutable)
print("scikit-learn     :", version_sklearn)

if ".venv" in ejecutable or "venv" in ejecutable:
    print("\nOK: el kernel apunta al entorno virtual del proyecto.")
else:
    print(
        "\nATENCION: el interprete no parece ser el .venv del proyecto."
        "\nCambia el kernel arriba a la derecha antes de seguir."
    )
````

## TODO 2 · Resolver la raíz del proyecto

Ninguna ruta absoluta escrita a mano en un notebook sobrevive a un cambio de carpeta.
La alternativa es calcular la raíz en tiempo de ejecución.

````python
# === TODO 2 — Encontrar la raiz del proyecto sin escribirla a mano ===
#
# Que hay que hacer:
#   Partiendo del directorio actual, subir de carpeta en carpeta hasta encontrar
#   una que contenga el archivo `requirements.txt`, y guardarla en RAIZ.
#
# Por que:
#   requirements.txt solo existe en la raiz del proyecto, asi que sirve de marca.
#   Con esto el notebook funciona sin cambios en Windows, macOS y Linux, lo abra
#   quien lo abra y este donde este la carpeta. Una ruta tipo
#   "C:/Users/.../ScikitLearn_2026/data" se rompe en cuanto mueves el proyecto.
#
# Que debe quedar:
#   RAIZ      -> Path, la carpeta que contiene requirements.txt
#   RUTA_DATA -> Path, la subcarpeta data/ de esa raiz
#
# Pistas: Path.cwd() es el directorio actual; .parent sube un nivel;
#         .exists() comprueba si algo existe; el operador / concatena rutas.
#         Un bucle while o un for sobre Path.cwd().parents sirven igual.

from pathlib import Path

# --- tu codigo aqui ---
RAIZ = Path.cwd()
while not (RAIZ / "requirements.txt").exists():
    if RAIZ == RAIZ.parent:          # ya estoy en la raíz del disco
        raise FileNotFoundError("No se encontró requirements.txt en ninguna carpeta padre")
    RAIZ = RAIZ.parent

RUTA_DATA = RAIZ / "data"


# --- verificacion (no modificar) ---
assert isinstance(RAIZ, Path), "RAIZ debe ser un objeto Path"
assert (RAIZ / "requirements.txt").exists(), f"No hay requirementLs.txt en {RAIZ}"
assert (RAIZ / "src" / "utils.py").exists(), f"No se ve src/utils.py bajo {RAIZ}"
assert RUTA_DATA == RAIZ / "data", "RUTA_DATA debe apuntar a la carpeta data/"

print("raiz del proyecto:", RAIZ)
print("carpeta de datos :", RUTA_DATA)
print("existe data/     :", RUTA_DATA.exists())
````

## TODO 4 · Cargar iris y explorar el `Bunch`

`load_iris()` no devuelve una tupla ni un DataFrame: devuelve un `Bunch`, un diccionario
cuyas claves también funcionan como atributos. Es el formato de todos los datasets
integrados, así que conviene conocerlo bien desde el primer día.

````python
# === TODO 4 — Cargar iris y extraer sus piezas ===
#
# Que hay que hacer:
#   a) Cargar el dataset iris completo (el Bunch, no solo los arrays) en `iris`.
#   b) Extraer la matriz de atributos en X y el vector objetivo en y.
#   c) Guardar en `nombres_clases` la lista de nombres de especie.
#   d) Guardar en `nombres_atributos` la lista de nombres de columna.
#
# Por que:
#   Todo modelo de scikit-learn recibe X con forma (n_muestras, n_atributos) y una
#   y con forma (n_muestras,). Reconocer esas dos formas de un vistazo te ahorra
#   la mitad de los errores de los proximos modulos.
#
# Que debe quedar:
#   iris              -> Bunch
#   X                 -> ndarray (150, 4)
#   y                 -> ndarray (150,)
#   nombres_clases    -> 3 nombres de especie
#   nombres_atributos -> 4 nombres de medida
#
# Pistas: load_iris() sin argumentos devuelve el Bunch.
#         Explora iris.keys() antes de escribir nada.

from sklearn.datasets import load_iris

# --- tu codigo aqui ---
iris = load_iris()
X = iris.data
y = iris.target
nombres_clases = iris.target_names
nombres_atributos = iris.feature_names


# --- verificacion (no modificar) ---
import numpy as np

assert X.shape == (150, 4), f"X deberia ser (150, 4) y es {X.shape}"
assert y.shape == (150,), f"y deberia ser (150,) y es {y.shape}"
assert len(nombres_clases) == 3, "Deberian ser 3 especies"
assert len(nombres_atributos) == 4, "Deberian ser 4 atributos"
assert set(np.unique(y)) == {0, 1, 2}, "y deberia contener las clases 0, 1 y 2"

resumen_bunch(iris, "iris")
print()
print("primera muestra :", X[0], "-> clase", y[0], f"({nombres_clases[y[0]]})")
````

print X_train.shape, y_train.shape

Antes de escribir: identifica las dos líneas que fallan y por qué.
Si dudas, la respuesta está en `docs/deprecaciones.md`, secciones 1 y 2.

````python
# === TODO 6 — Traducir el fragmento a Python 3 + scikit-learn moderno ===
#
# Que hay que hacer:
#   Reescribir el fragmento de arriba para que corra hoy, manteniendo exactamente
#   el mismo comportamiento: dos primeros atributos, 25% de test, random_state=33,
#   y el escalado ajustado SOLO sobre el conjunto de entrenamiento.
#
# Por que:
#   Las dos correcciones (el modulo que cambio de nombre y el print) son
#   mecanicas. La tercera cosa, que el libro ya hacia bien, es la importante:
#   fit sobre train, transform sobre ambos. Si ajustas el scaler sobre todo X,
#   la media y la desviacion del test se filtran al entrenamiento -> data leakage,
#   y tu metrica sale mejor de lo que sera en la realidad.
#
# Que debe quedar:
#   X_train_s, X_test_s -> ndarray escalados, (112, 2) y (38, 2)
#   y_train, y_test     -> ndarray (112,) y (38,)
#   media de X_train_s  -> practicamente 0
#   desviacion          -> practicamente 1
#
# Pistas: train_test_split ya no vive en sklearn.cross_validation.
#         Usa X_dos = X[:, :2] a partir de la X del TODO 4.
#         fit_transform en train, transform en test.

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

X_dos = X[:, :2]

# --- tu codigo aqui ---
X_train, X_test, y_train, y_test = train_test_split(
    X_dos, y, test_size=0.25, random_state=33
)
scaler = StandardScaler()

X_train_s = scaler. fit_transform(X_train)
X_test_s = scaler.transform(X_test)



# --- verificacion (no modificar) ---
import numpy as np

assert X_train.shape == (112, 2), f"X_train deberia ser (112, 2) y es {X_train.shape}"
assert X_test.shape == (38, 2), f"X_test deberia ser (38, 2) y es {X_test.shape}"
assert y_train.shape == (112,) and y_test.shape == (38,), "Formas de y incorrectas"

assert np.allclose(X_train_s.mean(axis=0), 0, atol=1e-9), (
    "La media de X_train_s deberia ser ~0 en cada columna. "
    "Ajustaste el scaler sobre el conjunto correcto?"
)
assert np.allclose(X_train_s.std(axis=0), 1, atol=1e-9), (
    "La desviacion de X_train_s deberia ser ~1 en cada columna."
)
assert not np.allclose(X_test_s.mean(axis=0), 0, atol=1e-6), (
    "La media de X_test_s ha salido exactamente 0: eso significa que ajustaste "
    "el scaler tambien sobre el test. Eso es data leakage."
)

print("formas   :", X_train_s.shape, X_test_s.shape)
print("train -> media", X_train_s.mean(axis=0).round(12), "desv", X_train_s.std(axis=0).round(12))
print("test  -> media", X_test_s.mean(axis=0).round(4), "desv", X_test_s.std(axis=0).round(4))
print("\nLa media del test NO es 0, y eso esta bien: se escalo con la media del train.")
````
